*   Error bars should represent 90% confidence intervals
*   The error bars should have whiskers that have a length of 0.05

The graph you produce should look idential to this:

Тепер, коли ми отримали графік, яким можна поділитися, давайте налаштуємо його відповідно до наступних  специфікацій, щоб відпрацювати свої навички Seaborn:

image.png



```python
# Create the bar plot with error bars
ax = sns.barplot(data=lab2_data, x='pclass', y='survived', hue='who', errorbar=('ci', 90), capsize=0.05)

# Define the new labels
current_labels = [tick.get_text() for tick in ax.get_xticklabels()]
new_labels = ['First Class', 'Second Class', 'Third Class']
ax.set_xticklabels(new_labels)

# Set the x-label
plt.xlabel('class')

# Move the legend to the center
legend = plt.legend(loc='center', bbox_to_anchor=(0.5, 0.5), ncol=1, title='who')
legend.get_title().set_fontsize('10')
plt.tight_layout()

plt.show()
```

Now that we've again obtained a graph we might want to share, let's adjust it to the following arbritrary specifications for practice:
*   Error bars should represent standard deviation
*   The error bars should have whiskers that have a length of 0.05
*   Instead of the sample mean, the bars should represent the sample median

*Hint: The third specification requires using NumPy.*

With these three changes, your graph should now look like this:


Тепер, коли ми знову отримали графік, яким, можливо, захочемо поділитися, давайте налаштуємо його відповідно до наступних довільних специфікацій для практики:
* Лінії похибок мають відображати стандартне відхилення
* Лінії помилок повинні мати вуса довжиною 0,05
* Замість вибіркового середнього стовпчики мають відображати медіану

*Підказка: третя специфікація вимагає використання NumPy.*

image.png

```python
import numpy as np

# Create the bar plot with error bars and median as the estimator
sns.barplot(data=lab2_data, x='survived_str', y='age', hue='who', errorbar='sd', capsize=0.05, estimator=np.nanmedian)

# Set the x-label
plt.xlabel('survived')

plt.show()
```

While this bar plot is now complete, there are better ways to present data as was discussed in the lecture. So to prouduce a graph more worthy of publication, let's convert our bar plot into a box plot and a strip plot.

Стовпчаста діаграма - готова, але існують кращі способи представлення даних, які ми вказали в лекції. Щоб створити графік для публікації в науковому журналі, давайте перетворимо наш стовпчастий графік на коробковий та точковий.

Since the inclusion of outliers in this graph does not change its scale much we can leave them in. However, if you data contains extreme outliers, remember you can remove them by passing the 'fliersize=0' argument to sns.boxplot() .

Оскільки включення викидів(outliers) у цей графік не сильно змінює його масштаб, ми можемо залишити їх. Якщо ваші дані містять екстремальні викиди, пам’ятайте, що ви можете видалити їх, передавши аргумент «fliersize=0» у sns.boxplot() .

```python
# Create the box plot
sns.boxplot(data=lab2_data, x='survived_str', y='age', hue='who')

# Set the x-label
plt.xlabel('survived')

plt.show()
```

Next, let's produce a strip plot version of this graph with datapoints that are half-transparent and jitter set to 0.3 .

Since the position of datapoints on strip plots are randomized, your graph should look similar, but not exactly like this:


Далі, створимо версію цього графіка з напівпрозорими точками даних і параметром jitter встановленим на 0,3.

Оскільки положення точок даних у цих графіках довільне, ваш графік має виглядати подібно, але на на 100%:

image.png

```python
# Create the strip plot
sns.stripplot(data=lab2_data, x='survived_str', y='age', hue='who', jitter=.3, alpha=.6, dodge=True)

# Set the x-label
plt.xlabel('survived')

plt.show()
```
